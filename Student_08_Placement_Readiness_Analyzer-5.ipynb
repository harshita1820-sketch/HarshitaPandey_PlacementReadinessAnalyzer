## 1. Import libraries
NumPy is used for numerical calculations. Pandas is used to read, organise, filter and analyse table data.


```python
import numpy as np
import pandas as pd
```

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
df = pd.read_csv("Student_08_dataset.csv")
print("Dataset loaded succesfully")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
# Display the number of rows and columns
print("Shape of dataset:", df.shape)

# Display column names
print("Column names:", df.columns)

# Display information about the dataset
df.info()

# Check missing values
print("Missing values:")
print(df.isnull().sum())

# Count duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# Select the four placement assessment columns
marks = df[['Aptitude', 'Reasoning', 'Verbal', 'Technical']]

# Convert the columns into a NumPy array
marks_array = marks.to_numpy()

# Calculate average, maximum and minimum
print("Average marks:", np.mean(marks_array))

print("Maximum marks:", np.max(marks_array))

print("Minimum marks:", np.min(marks_array))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# Create total marks column
df['Total_Marks'] = df[['Aptitude', 'Reasoning', 'Verbal', 'Technical']].sum(axis=1)

# Create overall average column
df['Overall_Average'] = df[['Aptitude', 'Reasoning', 'Verbal', 'Technical']].mean(axis=1)

# Filter students having average 70 or above
high_performers = df[df['Overall_Average'] >= 70]

print("Students with average 70 or above:")
display(high_performers)

# Sort students by overall average
top_students = df.sort_values(by='Overall_Average', ascending=False)

print("Top 5 students:")
display(top_students.head())

print("Bottom 5 students:")
display(top_students.tail())
```

## 6. Your individual project challenge
**Challenge:** Calculate overall average; apply a minimum 50 in every section rule; list eligible candidates.

Write your solution below and explain what the result means.


```python
# Check whether students scored at least 50 in every section
df['Placement_Eligible'] = (
    (df['Aptitude'] >= 50) &
    (df['Reasoning'] >= 50) &
    (df['Verbal'] >= 50) &
    (df['Technical'] >= 50)
)

# Display eligible students
eligible_students = df[df['Placement_Eligible'] == True]

print("Eligible Placement Candidates:")
display(eligible_students[['Student_ID', 'Name', 'Overall_Average']])

# Count eligible students
print("Total eligible students:", len(eligible_students))
```

## 7. Export your report
Use `to_csv(..., index=False)` to save your cleaned or analysed DataFrame. Create an `output` folder if needed.


```python
# Save the final placement readiness report
df.to_csv("Student_08_Placement_Readiness_Report.csv", index=False)

print("Final report saved successfully!")
```
